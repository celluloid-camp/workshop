# Download a YouTube video

This notebook shows how to download a YouTube video with [`yt-dlp`](https://github.com/yt-dlp/yt-dlp).

**Requirements:** Python 3.10+, and `ffmpeg` installed on your system for best format support.

## 2. Configure URL and output folder

```python
from pathlib import Path

# Replace with any public YouTube URL
YOUTUBE_URL = "https://www.youtube.com/watch?v=jNQXAC9IVRw"

OUTPUT_DIR = Path("downloads")
OUTPUT_DIR.mkdir(exist_ok=True)

print(f"Saving files to: {OUTPUT_DIR.resolve()}")
```

## 3. Download the video

Best effort: highest quality video + audio, merged into an mp4 when possible.

```python
import yt_dlp

ydl_opts = {
    "outtmpl": str(OUTPUT_DIR / "%(title)s [%(id)s].%(ext)s"),
    "format": "bv*+ba/b",
    "merge_output_format": "mp4",
    "noplaylist": True,
    "quiet": False,
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(YOUTUBE_URL, download=True)
    filepath = ydl.prepare_filename(info)

print(f"Downloaded: {filepath}")
```

## 5. List downloaded files

```python
for path in sorted(OUTPUT_DIR.iterdir()):
    if path.is_file():
        size_mb = path.stat().st_size / (1024 * 1024)
        print(f"{path.name} ({size_mb:.1f} MB)")
```
